```python
print("01. Add two numbers")

def add():
    print("This code is designed to add two numbers")
    num1 = int(input("Enter your first number: "))
    num2 = int(input("Enter your second number: "))
    total = int(num1+num2)
    print(f"The addition of two numbers, {num1} & {num2}, is: {total}")


# This provided line is required at the end of Python file to call the function.
if __name__ == '__main__':
    add()
```

```python
print("03. Fahrenheit to Celcius")

def main():
    print("This code is designed to convert Fahrenheit to Celcius")
    fahrenheit_degree = int(input("Enter the temperature in Fahrenheit: "))
    degrees_celsius = (fahrenheit_degree - 32) * 5.0/9.0
    print(f"The temperature in Celcius is: {degrees_celsius}")


# This provided line is required at the end of
# Python file to call the main() function.
if __name__ == '__main__':
    main()
```

```python
print("04. Age Riddle")

def age():
    print("This code is designed to calculate the ages of Anton, Beth, Chen, Drew, and Ethan")
    anton : int = 21  # Anton's age is given as 21 years old
    beth : int = 6 + anton  # Beth is 6 years older than Anton, so add 6 to Anton's age to get Beth's
    chen : int = 20 + beth  # Chen is 20 years older than Beth, so add 20 to Beth's age to get Chen's
    drew  : int= chen + anton  # Drew is as old as Chen's age plus Anton's age, so add them together
    ethan : int = chen  # Ethan is the same age as Chen, so set Ethan's age equal to Chen's

   # Print out all of the ages!
    print("Anton is " + str(anton)+ " years old.")
    print("Beth is " + str(beth)+ " years old.")
    print("Chen is " + str(chen)+ " years old.")
    print("Drew is " + str(drew)+ " years old.")
    print("Ethan is " + str(ethan)+ " years old.")


# There is no need to edit code beyond this point

if __name__ == '__main__':
    age()
```

```python
print("05. Perimeter of triangle")

def perimeter():
    print("This code is designed to calculate the perimeter of a triangle")
    side1: float = float(input("What is the length of side 1? "))
    side2: float = float(input("What is the length of side 2? "))
    side3: float = float(input("What is the length of side 3? "))
    print(f"The perimeter of the triangle is " + str(side1 + side2 + side3))

if __name__ == '__main__':
    perimeter()
```

```python
print("06. Squared number")

def square():
    num: float = float(input("Type a number to see its square: "))
    print(str(num) + " squared is " + str(num ** 2))

if __name__ == '__main__':
    square()
```
