```python
print("01. Dice Simulator")

import random

def roll_dice():
    """Simulates rolling two dice and prints their individual and total values."""
    dice1 = random.randint(1, 6)  # Generate a random number between 1 and 6 for the first die
    dice2 = random.randint(1, 6)  # Generate a random number between 1 and 6 for the second die
    total = dice1 + dice2  # Calculate the total of the two dice
    print(f"Die 1: {dice1}, Die 2: {dice2}, Total: {total}")  # Print the results

def main():
    """Calls the roll_dice function three times to simulate three rolls."""
    num_rolls = 3
    print(f"Simulating {num_rolls} dice rolls:")
    for _ in range(num_rolls):  # Loop three times
        roll_dice()  # Call the roll_dice function to perform a roll

if __name__ == "__main__":
    main()  # Start the program by calling the main function
```

```python
print("02. e-mc**2")


c:float = 299792458  # The speed of light in m/s

def calc_e_mc_squared():
  mass_in_kg= float(input("Enter the mass in kg: "))

  energy_in_joules= mass_in_kg * (c**2)

  print("The energy in joules is:", "{:.2f}".format(energy_in_joules), "J")

calc_e_mc_squared()
```

```python
print("05. Remainder Division")

def remainder_division():
    # Get the numbers we want to divide
    dividend: int = float(input("Please enter an integer to be divided: "))
    divisor: int = float(input("Please enter an integer to divide by: "))

    quotient: int = dividend // divisor  # Divide with no remainder/decimals (integer division)
    remainder: int = dividend % divisor  # Get the remainder of the division (modulo)

    print("The result of this division is " + str(quotient) + " with a remainder of " + str(remainder))


if __name__ == '__main__':
    remainder_division()
```

```python
print("07. Seconds in a year")

days_per_year: int = 365
hours_per_day: int = 24
minutes_per_hour: int = 60
seconds_per_minute: int = 60

def seconds_in_a_year():
    print("")
    seconds_in_a_year: int = days_per_year * hours_per_day * minutes_per_hour * seconds_per_minute
    print("There are", seconds_in_a_year, "seconds in a year.")

if __name__ == '__main__':
    seconds_in_a_year()
```
